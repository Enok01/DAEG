## Exercice1
Proposez un algorithme en python qui respecte les consignes suivantes:

- Demander à l’utilisateur un chiffre de 0 à 9 inclusivement
- Le programme doit afficher le nombre d’étoiles correspondant au chiffre inscrit. Eviter les retours lignes avec : print('*', end='') 
- Si la valeur inscrite par l’utilisateur n’est pas dans la plage, aucune étoile n’est affichée.
- Le programme se termine uniquement lorsque l’utilisateur a choisi deux fois le chiffre 0.  


```python
def affichage(nb_precedant=None):
    try:
        chiffre = int(input("Entrez un chiffre entre 0 et 9 inclusivement : "))

        if 0 <= chiffre <= 9:
            for _ in range(chiffre):
                print("*", end="")

            if chiffre == 0 and nb_precedant == 0:
                return  # sortie
            affichage(chiffre)

        else:
            print(f"{chiffre} n'est pas compris entre 0 et 9 !")
            affichage(nb_precedant)

    except ValueError:
        print("La valeur saisie n'est pas un nombre.")
        affichage(nb_precedant)

# appel de la fonction
affichage()







```

## Exercice 2
Ecrire un algorithme qui demande successivement 10 nombres à l’utilisateur, et qui lui dise ensuite quel était le plus grand parmi ces 10 nombres :

Entrez le nombre numéro 1 : 12
Entrez le nombre numéro 2 : 14
etc.
Entrez le nombre numéro 10 : 6
Le plus grand de ces nombres est : 14

On n'utilisera pas de tableau ici, ni de liste.
Modifiez ensuite l’algorithme pour que le programme affiche de surcroît en quelle position avait été saisie ce nombre


```python
i, plus_grand, position = 1, None, None

while i <= 10:
    try:
        nombre = int(input(f"Entrez le nombre numéro {i} : "))
        if plus_grand is None or nombre > plus_grand:
            plus_grand, position = nombre, i
        i += 1
    except ValueError:
        print("Saisie invalide, recommencez.")

print(f"Le plus grand nombre est {plus_grand} en position {position}.")



```

## Exercice 4 : États de l’eau en fonction de la température
Écrire une fonction etat_eau(temp) qui prend une température en degrés Celsius et retourne l’état de l’eau à cette température :
- "solide" si la température est inférieure à 0°C
- "liquide" si la température est entre 0°C et 100°C (inclus)
- "gazeux" si la température est strictement supérieure à 100°C

Utilisez un match/case pour cet exercice
Testez votre fonction en l'appelant

```python
def fonction etat_eau(temp) -> (str):
    try:
        match temp:
            case temp < 0:
                return "solide"
            case temp >=0 and temp<= 100:
                return "liquide"
            case _:
                return "gazeux"
    except ValueError:
        print("La valeur que vous avez rentrer n'est pas un nombre ")
    
```
