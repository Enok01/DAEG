# Laboratoire 2B (2h) — « Mathématiques et collections »
## Exercice 1 : Statistiques sur une liste de notes

Écris un programme qui :
- Génère une liste de 20 notes [aléatoires](https://www.w3schools.com/python/ref_random_randint.asp) entre 0 et 100
- Calcule la moyenne, la médiane, la note maximale et minimale
- Affiche ces résultats


## Exercice 2 : Dictionnaire de fréquences de notes

Reprends la liste de notes et crée un dictionnaire contenant le nombre d’occurrences de chaque note arrondie à la dizaine.

## Exercice 3 : Nombres premiers dans une liste

Génère une liste de 50 nombres entre 1 et 200, puis crée une nouvelle liste ne contenant que les nombres premiers.

## Exercice 4: Multiples communs entre deux listes

- Générez deux listes de 10 entiers aléatoires entre 1 et 50.
- Identifiez les nombres communs entre les deux listes (intersection) en utilisant une compréhension de liste
- S’il y a des nombres communs, trouvez tous les diviseurs entiers du plus grand nombre parmi ces communs.
- Affichez les résultats de façon claire.
