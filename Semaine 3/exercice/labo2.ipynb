## Exercice 1
Écrire un programme qui compte la fréquence des symboles textuels dans une liste de char à deux dimensions de 5*5 cases, puis enregistre les résultats dans une map: les symboles textuels seront en clef et leur fréquence respective sera la valeur. Votre programme doit afficher en sortie chaque symbole distinct et leur fréquence.

Exemple : image.png

```python
# Définir un tableau de caractères 5x5 
tableau = [
    ['@', 'X', '@', '?', '?'],
    ['&', 'X', 'X', '&', '&'],
    ['@', '!', '!', '?', '!'],
    ['!', 'X', '@', '&', 'X'],
    ['@', '&', 'X', '?', '!']
]
# continuer
def compter_caracteres(tableau: list) -> dict:
    compter = {}
    #compter = {key_instruction:value_instruction for key in dict if condition}
    for element in tableau:
        for caractere in element:
            if caractere in compter:
                compter[caractere] += 1
            else:
                compter[caractere] = 1
    return compter

print(compter_caracteres(tableau))
```
